```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the penguins.csv file into a DataFrame
df_penguins = pd.read_csv('/content/penguins.csv')

# Display the first 5 rows of the DataFrame
display(df_penguins.head())
```

```python
# List the column names and their data types
display(df_penguins.info())
```

```python
display(df_penguins.describe())
print(df_penguins.info())
print(df_penguins["species"].value_counts())
print(df_penguins.isna().sum())
```

```python
sns.histplot(data=df_penguins, x="body_mass_g", bins=20)
plt.title("Distribution of penguin Body Mass")
plt.show()

sns.histplot(data=df_penguins, x="body_mass_g", hue="species", bins=20)
plt.title("Distribution of penguin Body Mass")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_penguins, x="bill_length_mm", y="bill_depth_mm", hue="species", palette='viridis', s=80)
plt.title("Relationship between Bill Length and Bill Depth by Species")
plt.xlabel("Bill Length (mm)")
plt.ylabel("Bill Depth (mm)")
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(title="Species")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_penguins, x="species", y="body_mass_g", palette='viridis')
plt.title("Body Mass Distribution Across Penguin Species")
plt.xlabel("Species")
plt.ylabel("Body Mass (g)")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df_penguins.select_dtypes(include=['float64', 'int64']).corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap of Numeric Variables")
plt.show()
```

```python
sns.pairplot(df_penguins, hue='species', diag_kind='kde')
plt.suptitle('Pair Plot of Penguin Numeric Variables by Species', y=1.02) # Add a title above the plots
plt.show()
```



1.  **Why might Seaborn be preferred over plain Matplotlib for statistical visualization?**
    Seaborn is often preferred over plain Matplotlib for statistical visualization because it provides a higher-level interface for drawing attractive and informative statistical graphics. It builds on Matplotlib and integrates closely with Pandas data structures, making it easier to work with dataframes directly. Seaborn offers specialized plot types for common statistical tasks (like pair plots, heatmaps, box plots, etc.), sensible default aesthetics, and the ability to automatically handle complex plotting logic (like coloring by `hue` or faceting data), which would require significantly more code in plain Matplotlib. This often results in more concise code and visually appealing, statistically relevant plots with less effort.

2.  **What is the difference between a histogram and a box plot in terms of what they show about a distribution?**
    *   **Histogram:** A histogram displays the distribution of a single numeric variable. It groups data into bins and shows the frequency or count of data points falling into each bin. This allows you to see the shape of the distribution (e.g., symmetric, skewed), identify modes (peaks), and understand the spread and range of the data. For example, our `body_mass_g` histogram showed that the data is multimodal, especially when broken down by species.
    *   **Box Plot:** A box plot (or box-and-whisker plot) also displays the distribution of a single numeric variable, but it primarily focuses on summarizing its five-number summary: minimum, first quartile (Q1), median (Q2), third quartile (Q3), and maximum. It also highlights potential outliers. Box plots are excellent for comparing distributions across different categories, as we did with `body_mass_g` across `species`. While they don't show the exact shape of the distribution as clearly as a histogram, they provide a concise view of central tendency, spread, and skewness, and are very effective for comparative analysis.

3.  **Choose one plot from Part 3 and explain how it could help a researcher tell the three penguin species apart.**
    The **Pair Plot of Penguin Numeric Variables by Species** is incredibly effective for distinguishing the three penguin species. Specifically, the scatter plot showing the relationship between **Flipper Length (mm)** and **Body Mass (g)** (or vice-versa) clearly separates the species:
    *   **Gentoo** penguins form a very distinct cluster at the high end of both flipper length and body mass.
    *   **Adelie** penguins form another clear cluster with shorter flipper lengths and lower body masses.
    *   **Chinstrap** penguins are situated between Adelie and Gentoo, with intermediate flipper lengths and body masses. Although there's some slight proximity to Adelie at the edges, the Chinstrap cluster is generally well-separated from both other species.
    This specific bivariate relationship acts as a strong discriminator, allowing a researcher to accurately classify penguins into their respective species based on these two physical measurements.

4.  **What is one limitation of the Palmer Penguins dataset, or one question you could not answer using it?**
    One significant limitation of the Palmer Penguins dataset is the **limited time frame of the data collection (3 years: 2007, 2008, 2009)**. This makes it challenging to analyze long-term trends or cyclical patterns that might occur over longer periods (e.g., decadal changes in population, climate impacts over multiple years).
    
    **One question we could not answer using this dataset is:** "What is the long-term impact of climate change on the body mass and breeding success of different penguin species in the Antarctic region?" To answer this, we would need data spanning several decades, including environmental variables, and potentially reproductive success metrics, which are not present or sufficiently long-term in this dataset.
